```python
import nltk
import re
import string
from nltk.corpus import stopwords, brown
from nltk.tokenize import word_tokenize, sent_tokenize, RegexpTokenizer
from nltk.stem import WordNetLemmatizer
```

```python
tafaseer1_text = "\n".join(tafaseer1)
```

```python
tafaseer2_text = "\n".join(tafaseer2)
```

```python
lower_tafaseer1_text = tafaseer1_text.lower()
lower_tafaseer2_text = tafaseer2_text.lower()
```

```python
def clean_text(lower_case):
    # split text phrases into words
    words  = nltk.word_tokenize(lower_case)
    
    
    # Create a list of all the punctuations we wish to remove
    punctuations = ['.', ',', '/', '!', '?', ';', ':', '(',')', '[',']', '-', '_', '%']
    
    # Remove all the special characters
    punctuations = re.sub(r'\W', ' ', str(lower_case))
    
    stop_words  = stopwords.words('english')
    
    # Getting rid of all the words that contain numbers in them
    w_num = re.sub('\w*\d\w*', '', lower_case).strip()
    
    # remove all single characters
    lower_case = re.sub(r'\s+[a-zA-Z]\s+', ' ', lower_case)
    
    # Substituting multiple spaces with single space
    lower_case = re.sub(r'\s+', ' ', lower_case, flags=re.I)
        
    # Removing non-english characters
    lower_case = re.sub(r'^b\s+', '', lower_case)
    
    # Return keywords which are not in stop words 
    keywords = [word for word in words if not word in stop_words  and word in punctuations and  word in w_num]
    
    return keywords
```

```python
# Lemmatize the words
wordnet_lemmatizer = WordNetLemmatizer()

lemmatized_word_tafseer = [wordnet_lemmatizer.lemmatize(word) for word in clean_text(lower_tafaseer1_text)]

clean_lower_tafaseer1_text = ' '.join(lemmatized_word_tafseer)
len(clean_lower_tafaseer1_text)
```

```python
# Lemmatize the words
wordnet_lemmatizer = WordNetLemmatizer()

lemmatized_word_tafseer = [wordnet_lemmatizer.lemmatize(word) for word in clean_text(lower_tafaseer2_text)]

clean_lower_tafaseer2_text = ' '.join(lemmatized_word_tafseer)
len(clean_lower_tafaseer2_text)
```

```python
import nltk
from nltk.corpus import stopwords
from collections import Counter
import matplotlib.pyplot as plt

# Make sure you have stopwords
nltk.download('stopwords')

# Let's say this is your tokenized list of words
# clean_lower_tafaseer1_text = ['allah', 'merciful', 'compassionate', 'allah', 'merciful']

# Step 1: Count words
word_count_dict = Counter(clean_lower_tafaseer1_text)

# Step 2: Remove stopwords
popular_words = sorted(word_count_dict, key=word_count_dict.get, reverse=True)
popular_words_nonstop = [w for w in popular_words if w not in stopwords.words("english")]

# Step 3: Show top 10
for word in popular_words_nonstop[:10]:
    print(f"{word}: {word_count_dict[word]}")
```
